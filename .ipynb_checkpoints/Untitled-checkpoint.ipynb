```python
import warnings
warnings.filterwarnings(action="ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_predict
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (accuracy_score, classification_report, confusion_matrix,
                             recall_score, precision_score, roc_auc_score, average_precision_score,
                             PrecisionRecallDisplay, RocCurveDisplay)

RANDOM_STATE = 123
TARGET = "churned_next_60d"
LTV, COUPON, SAVE_RATE = 3000, 150, 0.30   # business assumptions
```

```python
Customer_data = pd.read_csv("quickbasket_customer_data.csv")
```

```python
print(Customer_data["churned_next_60d"].value_counts())
print(f"Churn rate: {Customer_data['churned_next_60d'].mean():.1%}")
```

```python
Customer_data = Customer_data.drop_duplicates().reset_index(drop=True)
print(Customer_data.shape)
```

```python
# Outliers: ₹45,000-₹99,999 order values are data-entry errors -> missing (imputed later in the pipeline)
Customer_data.loc[Customer_data["avg_order_value_inr"] > 10000, "avg_order_value_inr"] = np.nan

# No ticket raised -> no resolution time, so 0 is correct here (keep your existing cell for this)
Customer_data["avg_ticket_resolution_hrs"] = Customer_data["avg_ticket_resolution_hrs"].fillna(0)

# Blank CSAT / app rating = "did not respond", so keep that as a flag instead of a fake 0
Customer_data["csat_missing"]       = Customer_data["last_csat_score"].isna().astype(int)
Customer_data["app_rating_missing"] = Customer_data["app_store_rating_given"].isna().astype(int)

# Small basket + high fee is a stronger pain signal than either number alone
Customer_data["fee_to_basket_ratio"] = Customer_data["avg_delivery_fee_inr"] / Customer_data["avg_order_value_inr"]
```

```python
cat_cols  = ["city", "warehouse_id", "gender", "acquisition_channel",
             "membership_plan", "preferred_payment", "top_category"]
drop_cols = ["customer_id", "signup_date", TARGET, "app_store_rating_given"]
feature_cols = [c for c in Customer_data.columns if c not in drop_cols]
num_cols     = [c for c in feature_cols if c not in cat_cols]

X = Customer_data[feature_cols]
y = Customer_data[TARGET]

# Split BEFORE any fitting. Everything else is learned from training data only.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=True, random_state=RANDOM_STATE, stratify=y)
print(X_train.shape, X_test.shape, len(num_cols), "numeric +", len(cat_cols), "categorical")
```

```python
def make_preprocessor(scale=True):
    num_steps = [("impute", SimpleImputer(strategy="median"))]
    if scale:
        num_steps.append(("scale", StandardScaler()))
    return ColumnTransformer([
        ("num", Pipeline(num_steps), num_cols),
        ("cat", Pipeline([("impute", SimpleImputer(strategy="constant", fill_value="Unknown")),
                          ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False))]), cat_cols)])

models = {
    "Logistic Regression": Pipeline([("pre", make_preprocessor(True)),
                                     ("clf", LogisticRegression(max_iter=3000))]),
    "Random Forest":       Pipeline([("pre", make_preprocessor(False)),
                                     ("clf", RandomForestClassifier(n_estimators=300, min_samples_leaf=10,
                                                                    n_jobs=-1, random_state=RANDOM_STATE))]),
    "Gradient Boosting":   Pipeline([("pre", make_preprocessor(False)),
                                     ("clf", GradientBoostingClassifier(n_estimators=200, learning_rate=0.05,
                                                                        max_depth=3, subsample=0.8,
                                                                        random_state=RANDOM_STATE))]),
}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

def net_benefit(y_true, p, thr):
    flag = np.asarray(p) >= thr
    yt = np.asarray(y_true)
    tp = (flag & (yt == 1)).sum()
    fp = (flag & (yt == 0)).sum()
    return tp * (SAVE_RATE * LTV - COUPON) - fp * COUPON

thresholds = np.arange(0.02, 0.95, 0.01)

oof, thr_by_model, rows = {}, {}, []
for name, pipe in models.items():
    # out-of-fold probabilities on TRAIN only (the test set stays untouched)
    oof[name] = cross_val_predict(pipe, X_train, y_train, cv=cv, method="predict_proba")[:, 1]
    nb = [net_benefit(y_train, oof[name], t) for t in thresholds]
    thr_by_model[name] = thresholds[int(np.argmax(nb))]
    rows.append({"model": name,
                 "CV ROC-AUC": roc_auc_score(y_train, oof[name]),
                 "CV PR-AUC": average_precision_score(y_train, oof[name]),
                 "best threshold": thr_by_model[name],
                 "CV net benefit ₹": max(nb)})
cv_table = pd.DataFrame(rows).set_index("model")
best_name = cv_table["CV net benefit ₹"].idxmax()
print("Best model:", best_name)
cv_table.round(3)
```

```python
# Fit each model ONCE and get probabilities (not hard 0/1 labels)
fitted  = {name: pipe.fit(X_train, y_train) for name, pipe in models.items()}
p_test  = {name: m.predict_proba(X_test)[:, 1] for name, m in fitted.items()}

def evaluate_model(name):
    p, thr = p_test[name], thr_by_model[name]
    print(f"{name}: ROC-AUC = {roc_auc_score(y_test, p):.3f} | PR-AUC = {average_precision_score(y_test, p):.3f} (chance = {y_test.mean():.3f})")
    for label, t in [("Default threshold 0.50", 0.5), (f"Cost-optimal threshold {thr:.2f}", thr)]:
        pred = (p >= t).astype(int)
        print(f"\n--- {label} ---")
        print(classification_report(y_test, pred, digits=2))
        print(confusion_matrix(y_test, pred))
```

```python
evaluate_model("Logistic Regression")   # under #### LogisticRegression
evaluate_model("Random Forest")         # under #### Random Forest
evaluate_model("Gradient Boosting")     # under #### GradientBoostingClassifier
```

```python
fig, ax = plt.subplots(1, 2, figsize=(13, 5))
for name in models:
    PrecisionRecallDisplay.from_predictions(y_test, p_test[name], name=name, ax=ax[0])
    RocCurveDisplay.from_predictions(y_test, p_test[name], name=name, ax=ax[1])
ax[0].axhline(y_test.mean(), color="k", ls="--", lw=1)   # chance level
ax[1].plot([0, 1], [0, 1], "k--", lw=1)
plt.tight_layout(); plt.show()
```

```python
final_name = best_name
p_final    = p_test[final_name]
final_thr  = thr_by_model[final_name]

def strategy_row(name, flag):
    flag = np.asarray(flag).astype(bool); yt = y_test.values
    tp = (flag & (yt == 1)).sum(); fp = (flag & (yt == 0)).sum()
    return {"strategy": name, "targeted": int(flag.sum()), "recall": tp / yt.sum(),
            "precision": tp / max(flag.sum(), 1),
            "net_benefit_per_10k_₹": int((tp*(SAVE_RATE*LTV-COUPON) - fp*COUPON) / len(yt) * 10000)}

pd.DataFrame([
    strategy_row("Do nothing", np.zeros(len(y_test))),
    strategy_row("Offer everyone", np.ones(len(y_test))),
    strategy_row("Model @ 0.5", p_final >= 0.5),
    strategy_row(f"Model @ {final_thr:.2f}", p_final >= final_thr),
])
```

```python
# Logistic-regression coefficients: direction of each driver
lr = fitted["Logistic Regression"]
names = [n.split("__", 1)[1] for n in lr.named_steps["pre"].get_feature_names_out()]
coefs = pd.Series(lr.named_steps["clf"].coef_[0], index=names).sort_values()
coefs.tail(10)      # biggest risk-raisers
```

```python
# Segment evidence: which team's claim holds up?
df = Customer_data
for label, mask in {"DEL-02 warehouse": df.warehouse_id == "DEL-02",
                    "Delivery >= 40 min": df.avg_delivery_time_min >= 40,
                    "Ticket resolution >= 24h": df.avg_ticket_resolution_hrs >= 24,
                    "Discount >= 40%": df.first_order_discount_pct >= 40,
                    "Fee >= ₹39": df.avg_delivery_fee_inr >= 39}.items():
    print(f"{label:28s} churn inside {df.loc[mask, TARGET].mean():.0%} vs outside {df.loc[~mask, TARGET].mean():.0%}")
```
