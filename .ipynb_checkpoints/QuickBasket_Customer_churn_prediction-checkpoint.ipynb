## 4. Data Cleaning

```python
Customer_data.drop_duplicates(inplace=True) #Removing Duplicate Entries
```

```python
#After removing Duplicates
print("Customer data After removing duplicates, \n Record count :",Customer_data.shape[0],"\n Feature Count",Customer_data.shape[1])
```

```python
Customer_data.isna().sum() # Total number of null entries
print(Customer_data.isna().sum()[Customer_data.isna().sum() > 0])
```

```python
missing_summary = pd.DataFrame({
    "missing_count": Customer_data.isnull().sum(),
    "missing_percentage": (
        Customer_data.isnull().mean() * 100
    ).round(2)
})

print(missing_summary[missing_summary["missing_count"] > 0])
```

```python
print("Total null entries in age column before cleanup is",Customer_data["age"].isna().sum())
```

```python
Customer_data["age"] = Customer_data["age"].fillna(Customer_data["age"].median())
print("Total null entries in age column after cleanup is",Customer_data["age"].isna().sum())
```

```python
print("Total null entries in age column before cleanup is",Customer_data["gender"].isna().sum())
print("Genders in Data set before cleanup \n",Customer_data["gender"].value_counts())
```

```python
Customer_data["gender"] = Customer_data["gender"].fillna("UNKNOWN")
print("Total null entries in age column after cleanup is",Customer_data["gender"].isna().sum())
print("Genders in Data set after cleanup \n",Customer_data["gender"].value_counts())
```

```python
Customer_data.describe()
```

```python
Customer_data["avg_order_value_inr"].fillna(Customer_data["avg_order_value_inr"].median(),inplace=True)
```

```python
Customer_data.sort_values(by="avg_order_value_inr",ascending=True)
```

```python
print(Customer_data.isnull().sum()[Customer_data.isnull().sum() > 0])
```

## 5. Data Prepration

```python
X = Customer_data.drop(columns=["customer_id","signup_date","churned_next_60d"])
y = Customer_data["churned_next_60d"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,shuffle=True,random_state=123,stratify=y)
```
