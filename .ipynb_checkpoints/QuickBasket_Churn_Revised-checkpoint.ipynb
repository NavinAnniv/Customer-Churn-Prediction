# QuickBasket: corrected churn analysis

Read `QuickBasket_Churn_Report.pdf` and `README.md` first. The supplied future labels are unverified as of 16 October 2026; results are provisional. Run from this folder. All five charts and tabular results are provided alongside this notebook.

The cells below contain the complete analysis. Code execution counts are deliberately unset: the identical `analysis.py` was run as a script; its captured results and images appear at the end. Running all cells reproduces the analysis.

```python
import sys
sys.argv=['analysis.py','quickbasket_customer_data.csv']
```

## Data, preprocessing, cross-validation and economic policies

```python
"""QuickBasket reproducible analysis. Run: python analysis.py [csv_path].
All estimates are retrospective and conditional on supplied, unverified labels.
"""
from pathlib import Path
import sys, json, warnings
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_predict
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.calibration import CalibratedClassifierCV, calibration_curve
from sklearn.metrics import average_precision_score, roc_auc_score, brier_score_loss, precision_score, recall_score, confusion_matrix, PrecisionRecallDisplay
from sklearn.inspection import permutation_importance
warnings.filterwarnings('ignore', category=FutureWarning)
OUT=Path('deliverables'); OUT.mkdir(exist_ok=True)
plt.rcParams.update({'figure.dpi':140,'axes.spines.top':False,'axes.spines.right':False,'font.size':10})
SEED=123
path=Path(sys.argv[1]) if len(sys.argv)>1 else Path('upload/quickbasket_customer_data.csv')
raw=pd.read_csv(path); d=raw.drop_duplicates().copy()
assert d.customer_id.is_unique, 'Conflicting customer rows require reconciliation'
assert d.churned_next_60d.isin([0,1]).all()
y=d.churned_next_60d.astype(int)
# Exclude identity, redundant signup clock, and protected attributes from targeting.
drop=['customer_id','signup_date','churned_next_60d','age','gender']
X=d.drop(columns=drop)
num=X.select_dtypes(include='number').columns.tolist(); cat=X.select_dtypes(exclude='number').columns.tolist()
def clean_numeric(a):
    a=a.copy()
    # AOV > INR 5,000 is an explicit audit rule, independent of labels/splits.
    a.loc[a.avg_order_value_inr>5000,'avg_order_value_inr']=np.nan
    return a
def preprocessing(cols=None):
    nn=num if cols is None else [c for c in num if c in cols]
    cc=cat if cols is None else [c for c in cat if c in cols]
    return ColumnTransformer([
        ('numeric',Pipeline([('audit',FunctionTransformer(clean_numeric,feature_names_out='one-to-one')),('impute',SimpleImputer(strategy='median',add_indicator=True)),('scale',StandardScaler())]),nn),
        ('category',Pipeline([('impute',SimpleImputer(strategy='constant',fill_value='Unknown')),('encode',OneHotEncoder(handle_unknown='ignore',sparse_output=False))]),cc)
    ],verbose_feature_names_out=False)
def pipeline(model,cols=None): return Pipeline([('preprocess',preprocessing(cols)),('model',model)])
models={
 'Logistic':pipeline(LogisticRegression(max_iter=3000,C=1)),
 'Balanced logistic (calibrated)':CalibratedClassifierCV(pipeline(LogisticRegression(max_iter=3000,C=1,class_weight='balanced')),method='sigmoid',cv=3),
 'Random forest':pipeline(RandomForestClassifier(n_estimators=250,max_depth=7,min_samples_leaf=20,max_features=.8,random_state=SEED,n_jobs=1))
}
itr,ite=train_test_split(np.arange(len(d)),test_size=.2,random_state=SEED,stratify=y)
xt,xv=X.iloc[itr],X.iloc[ite]; yt,yv=y.iloc[itr],y.iloc[ite]
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=SEED)
oof={}; rows=[]
for name,m in models.items():
 print('Fitting',name,flush=True)
 p=cross_val_predict(m,xt,yt,cv=cv,method='predict_proba',n_jobs=1)[:,1]; oof[name]=p
 folds=[]
 for _,val in cv.split(xt,yt): folds.append(average_precision_score(yt.iloc[val],p[val]))
 rows.append({'model':name,'cv_average_precision':average_precision_score(yt,p),'cv_fold_ap_sd':np.std(folds,ddof=1),'cv_roc_auc':roc_auc_score(yt,p),'cv_brier':brier_score_loss(yt,p)})
scores=pd.DataFrame(rows).sort_values('cv_average_precision',ascending=False)
# Prefer simpler logistic if its AP is within 0.01 of the best model.
best=scores.iloc[0].model
if scores.set_index('model').loc['Logistic','cv_average_precision']>=scores.iloc[0].cv_average_precision-.01: best='Logistic'
print(scores.to_string(index=False), '\nSELECTED',best,flush=True)
def sweep(y,p):
 rows=[]
 for t in np.linspace(0,1,1001):
  z=p>=t; tn,fp,fn,tp=confusion_matrix(y,z,labels=[0,1]).ravel()
  rows.append({'threshold':t,'flagged':int(z.sum()),'tp':int(tp),'fp':int(fp),'fn':int(fn),'tn':int(tn),'precision':tp/max(1,tp+fp),'recall':tp/max(1,tp+fn),'weighted_error_cost':150*fp+3000*fn,'net_value_u_one_third':1000*tp-150*(tp+fp)})
 return pd.DataFrame(rows)
sw=sweep(yt,oof[best]); t=float(sw.loc[sw.net_value_u_one_third.idxmax(),'threshold']); terror=float(sw.loc[sw.weighted_error_cost.idxmin(),'threshold'])
sw.to_csv(OUT/'threshold_sweep_training_oof.csv',index=False)
fitted=clone(models[best]).fit(xt,yt); p=fitted.predict_proba(xv)[:,1]
tests=[]
for name,m in models.items():
 mm=fitted if name==best else clone(m).fit(xt,yt)
 pp=mm.predict_proba(xv)[:,1]
 tests.append({'model':name,'test_ap':average_precision_score(yv,pp),'test_roc_auc':roc_auc_score(yv,pp),'test_brier':brier_score_loss(yv,pp)})
scores=scores.merge(pd.DataFrame(tests),on='model');scores.to_csv(OUT/'model_comparison.csv',index=False)
policies=[]
for name,cut in [('Flag none',1.01),('Default 0.5',.5),('Analytic treatment 0.15',.15),('OOF treatment optimum',t),('Analytic error cost 0.0476',150/3150),('OOF error cost optimum',terror),('Flag all',0)]:
 s=sweep(yv,p) if False else None
 z=p>=cut; tn,fp,fn,tp=confusion_matrix(yv,z,labels=[0,1]).ravel()
 policies.append(dict(policy=name,threshold=cut,flagged=int(z.sum()),tp=int(tp),fp=int(fp),fn=int(fn),tn=int(tn),precision=tp/max(1,tp+fp),recall=tp/max(1,tp+fn),weighted_error_cost=150*fp+3000*fn,net_value_u_one_third=1000*tp-150*(tp+fp)))
pol=pd.DataFrame(policies);pol.to_csv(OUT/'test_policy_comparison.csv',index=False)
rng=np.random.default_rng(123);boot=[]
for _ in range(1500):
 ix=rng.integers(0,len(yv),len(yv));yy=yv.to_numpy()[ix];pp=p[ix]; zz=pp>=t
 boot.append([average_precision_score(yy,pp),roc_auc_score(yy,pp),recall_score(yy,zz),precision_score(yy,zz,zero_division=0),1000*np.sum(zz & (yy==1))-150*zz.sum()])
cis={n:np.quantile(np.array(boot)[:,i],[.025,.975]).tolist() for i,n in enumerate(['ap','auc','recall','precision','scenario_net_value'])}
```

## Logistic and marginal permutation explanations

```python
# Explanations: unweighted logistic for direction/OR and marginal holdout permutation for predictive dependence.
lr=clone(models['Logistic']).fit(xt,yt)
names=lr.named_steps['preprocess'].get_feature_names_out()
co=pd.DataFrame({'feature':names,'coefficient':lr.named_steps['model'].coef_[0]})
co['odds_ratio']=np.exp(co.coefficient);co['abs_coefficient']=co.coefficient.abs()
co.sort_values('abs_coefficient',ascending=False).to_csv(OUT/'logistic_coefficients.csv',index=False)
pi=permutation_importance(fitted,xv,yv,scoring='average_precision',n_repeats=15,random_state=123,n_jobs=1)
imp=pd.DataFrame({'feature':X.columns,'ap_drop':pi.importances_mean,'repeat_sd':pi.importances_std}).sort_values('ap_drop',ascending=False)
imp.to_csv(OUT/'driver_permutation.csv',index=False)
```

## Grouped permutation for correlated features

```python
# Jointly permute correlated service and engagement families, maintaining their internal rows.
groups={'Engagement':['days_since_last_order','orders_last_90d','app_sessions_per_week'], 'Operations':['avg_delivery_time_min','late_delivery_pct','order_cancellation_pct','missing_or_wrong_items_pct'], 'Support':['support_tickets_90d','avg_ticket_resolution_hrs'], 'Price and plan':['membership_plan','avg_delivery_fee_inr','avg_order_value_inr'], 'Acquisition':['acquisition_channel','first_order_discount_pct','coupon_usage_pct'], 'Satisfaction':['last_csat_score','app_store_rating_given'], 'Geography':['city','warehouse_id']}
gp=[]
for name,cols in groups.items():
 dif=[]
 for _ in range(20):
  xx=xv.copy();xx[cols]=xv.iloc[rng.permutation(len(xv))][cols].to_numpy()
  dif.append(average_precision_score(yv,p)-average_precision_score(yv,fitted.predict_proba(xx)[:,1]))
 gp.append({'group':name,'ap_drop':np.mean(dif),'repeat_sd':np.std(dif)})
pd.DataFrame(gp).sort_values('ap_drop',ascending=False).to_csv(OUT/'driver_groups.csv',index=False)
```

## Sensitivity to engagement fields

```python
# Leakage sensitivity uses training OOF only, preserving holdout.
ab=[]
for label,excluded in [('No recency',['days_since_last_order']),('No engagement',groups['Engagement'])]:
 cols=[c for c in X.columns if c not in excluded]
 a=pipeline(LogisticRegression(max_iter=3000),cols)
 pp=cross_val_predict(a,xt[cols],yt,cv=cv,method='predict_proba')[:,1]
 ab.append({'model':label,'cv_ap':average_precision_score(yt,pp),'cv_auc':roc_auc_score(yt,pp)})
pd.DataFrame(ab).to_csv(OUT/'leakage_sensitivity.csv',index=False)
```

## Descriptive segment rates and quality profiles

```python
# Full-cohort descriptive segment rates; Wilson intervals and sample sizes.
segments=[]
for col in ['city','warehouse_id','membership_plan','acquisition_channel']:
 for val,g in d.groupby(col):
  n=len(g);k=int(g.churned_next_60d.sum());q=k/n; z=1.96
  center=(q+z*z/(2*n))/(1+z*z/n);delta=z*np.sqrt(q*(1-q)/n+z*z/(4*n*n))/(1+z*z/n)
  segments.append({'dimension':col,'segment':val,'n':n,'churners':k,'rate':q,'lo':center-delta,'hi':center+delta})
seg=pd.DataFrame(segments);seg.to_csv(OUT/'segment_rates.csv',index=False)
profile=d.groupby('warehouse_id')[['churned_next_60d','avg_delivery_time_min','late_delivery_pct','order_cancellation_pct','missing_or_wrong_items_pct','support_tickets_90d','avg_ticket_resolution_hrs','avg_delivery_fee_inr','days_since_last_order']].mean()
profile.to_csv(OUT/'warehouse_profile.csv')
drivers=[]
for c,condition in [('days_since_last_order >=30',d.days_since_last_order>=30),('orders_last_90d <=3',d.orders_last_90d<=3),('late_delivery_pct >=25',d.late_delivery_pct>=25),('avg_delivery_fee_inr >=40',d.avg_delivery_fee_inr>=40),('ticket resolution >24h (ticketed only)',d.avg_ticket_resolution_hrs>24),('support tickets >=2',d.support_tickets_90d>=2),('first discount >=30',d.first_order_discount_pct>=30),('CSAT <=3 (responders)',d.last_csat_score<=3)]:
 eligible=pd.Series(True,index=d.index)
 if c.startswith('ticket resolution'):eligible=d.support_tickets_90d>0
 if c.startswith('CSAT'):eligible=d.last_csat_score.notna()
 drivers.append({'condition':c,'n_high':int((condition&eligible).sum()),'rate_high':y[condition&eligible].mean(),'n_other':int((~condition&eligible).sum()),'rate_other':y[~condition&eligible].mean()})
pd.DataFrame(drivers).to_csv(OUT/'business_driver_rates.csv',index=False)
```

## Capacity and treatment-response assumptions

```python
# Capacity and response sensitivity; cutoffs are training OOF-derived.
cap=[]
for frac in [.1,.2,.3,.5]:
 cutoff=np.quantile(oof[best],1-frac);z=p>=cutoff
 cap.append({'training_capacity':frac,'threshold':cutoff,'test_flagged':int(z.sum()),'test_recall':recall_score(yv,z),'test_precision':precision_score(yv,z),'net_value_u_one_third':1000*np.sum(z&(yv==1))-150*z.sum()})
pd.DataFrame(cap).to_csv(OUT/'capacity_scenarios.csv',index=False)
sens=[]
for u in [.1,.2,1/3,.5]:
 v=u*3000*sw.tp-150*sw.flagged;cut=float(sw.loc[v.idxmax(),'threshold']);zz=p>=cut
 sens.append({'rescue_rate':u,'analytic_threshold':150/(u*3000),'oof_threshold':cut,'test_flagged':int(zz.sum()),'test_net_value':float(u*3000*np.sum(zz&(yv==1))-150*zz.sum())})
pd.DataFrame(sens).to_csv(OUT/'treatment_sensitivity.csv',index=False)
```

```python
# Five deliverable charts.
fig,ax=plt.subplots(figsize=(8,5));PrecisionRecallDisplay.from_predictions(yv,p,ax=ax,name=best,plot_chance_level=True);ax.set_title('Held-out ranking of future churn (supplied labels)');fig.tight_layout();fig.savefig(OUT/'01_precision_recall.png');plt.close(fig)
fig,axs=plt.subplots(1,2,figsize=(11,4.4));axs[0].plot(sw.threshold,sw.weighted_error_cost/len(yt)*1000/100000);axs[0].axvline(terror,color='#db7a20',ls='--');axs[0].set(title='Weighted errors: FP 150, FN 3,000',ylabel='INR lakh per 1,000 customers',xlabel='Probability cutoff');axs[1].plot(sw.threshold,sw.net_value_u_one_third/len(yt)*1000/100000,color='#16847b');axs[1].axvline(t,color='#db7a20',ls='--',label=f'OOF optimum {t:.3f}');axs[1].axhline(0,color='gray',lw=.7);axs[1].set(title='Treatment: 1/3 rescue, INR 150 per flag',ylabel='Expected net value, lakh / 1,000',xlabel='Probability cutoff');axs[1].legend();fig.suptitle('Thresholds chosen using training out-of-fold scores only');fig.tight_layout();fig.savefig(OUT/'02_threshold_economics.png');plt.close(fig)
fig,axs=plt.subplots(1,2,figsize=(11,4.8));q=imp.head(10).iloc[::-1];axs[0].barh(q.feature,q.ap_drop,color='#16847b');axs[0].set(title='Individual predictors',xlabel='Held-out average precision loss');g=pd.DataFrame(gp).sort_values('ap_drop');axs[1].barh(g.group,g.ap_drop,color='#355c87');axs[1].set(title='Correlated feature families (joint permutation)',xlabel='Held-out average precision loss');fig.tight_layout();fig.savefig(OUT/'03_predictive_drivers.png');plt.close(fig)
fig,axs=plt.subplots(2,2,figsize=(11,8));
for ax,col in zip(axs.ravel(),['warehouse_id','city','acquisition_channel','membership_plan']):
 s=seg[seg.dimension==col].sort_values('rate');ax.barh(s.segment,s.rate*100,color=['#dc7e43' if x=='DEL-02' else '#355c87' for x in s.segment]);ax.errorbar(s.rate*100,np.arange(len(s)),xerr=np.array([s.rate-s.lo,s.hi-s.rate])*100,fmt='none',color='black',lw=1);ax.axvline(y.mean()*100,color='gray',ls='--');ax.set(title=col,xlabel='Churn % (95% Wilson interval)')
fig.suptitle('Observed concentration: descriptive, unadjusted rates');fig.tight_layout();fig.savefig(OUT/'04_segment_concentration.png');plt.close(fig)
fig,ax=plt.subplots(figsize=(7,4.6));a,b=calibration_curve(yv,p,n_bins=8,strategy='quantile');ax.plot(b,a,'o-',label='Holdout');a,b=calibration_curve(yt,oof[best],n_bins=8,strategy='quantile');ax.plot(b,a,'s--',label='Training OOF');ax.plot([0,.8],[0,.8],':',color='gray');ax.set(xlabel='Mean predicted churn probability',ylabel='Observed churn fraction',title='Calibration: needed for cost-based targeting');ax.legend();fig.tight_layout();fig.savefig(OUT/'05_calibration.png');plt.close(fig)
pred=pd.DataFrame({'customer_id':d.iloc[ite].customer_id,'actual_churn':yv,'predicted_probability':p,'flag_at_oof_threshold':p>=t});pred.to_csv(OUT/'holdout_predictions.csv',index=False)
fair=pred.merge(d[['customer_id','age','gender','city']],on='customer_id')
fair['age_band']=pd.cut(fair.age,[17,24,34,44,100]).astype(str)
fair_rows=[]
for col in ['gender','age_band','city']:
 for group,g in fair.groupby(col,dropna=False):
  zz=g.flag_at_oof_threshold; yy=g.actual_churn
  fair_rows.append(dict(dimension=col,group=group,n=len(g),churners=int(yy.sum()),recall=((yy==1)&zz).sum()/max(1,yy.sum()),false_positive_rate=((yy==0)&zz).sum()/max(1,(yy==0).sum())))
pd.DataFrame(fair_rows).to_csv(OUT/'subgroup_audit.csv',index=False)
audit={'raw_rows':len(raw),'unique_customers':len(d),'duplicates_removed':len(raw)-len(d),'churners':int(y.sum()),'churn_rate':y.mean(),'train_n':len(yt),'test_n':len(yv),'test_churners':int(yv.sum()),'aov_outliers_above_5000':int((d.avg_order_value_inr>5000).sum()),'missing':d.isna().sum().to_dict(),'recency_ge60':int((d.days_since_last_order>=60).sum()),'zero_orders':int((d.orders_last_90d==0).sum()),'signup_tenure_mismatch':int((((pd.Timestamp('2026-09-30')-pd.to_datetime(d.signup_date)).dt.days)!=d.tenure_days).sum()),'recency_gt_tenure':int((d.days_since_last_order>d.tenure_days).sum()),'selected_model':best,'threshold':t,'error_threshold':terror,'bootstrap_95_ci':cis,'top_groups':gp}
(OUT/'audit_and_results.json').write_text(json.dumps(audit,indent=2))
print('AUDIT',json.dumps(audit,indent=2));print(pol.to_string(index=False));print('DRIVERS',imp.head(12).to_string(index=False));print(profile.to_string());print(pd.DataFrame(drivers).to_string(index=False));print('DONE',flush=True)
```
