```python
Customer_data["gender"] = Customer_data["gender"].fillna("UNKNOWN")
print("Total null entries in age column after cleanup is",Customer_data["gender"].isna().sum())
print("Genders in Data set after cleanup \n",Customer_data["gender"].value_counts())
```

## 5. Data Prepration

```python
X = Customer_data.drop(columns=["customer_id","signup_date","churned_next_60d"])
y = Customer_data["churned_next_60d"]
```

```python
X.shape,y.shape
```

```python
columns = [
    "city", "warehouse_id", "gender", "acquisition_channel",
    "membership_plan", "preferred_payment", "top_category"
]

encoders = {}

for column in columns:
    encoders[column] = LabelEncoder()
    X[column] = encoders[column].fit_transform(X[column])
```

```python
scaler = StandardScaler()
```

```python
Scaled_X = scaler.fit_transform(X)
```

```python
lr_Model = LogisticRegression()
GB_Model = GradientBoostingClassifier()
RF_Model = RandomForestClassifier(n_estimators=100,max_depth=3)
```

```python
selector = SelectFromModel(estimator=RF_Model, threshold="mean")
X_selected = selector.fit_transform(Scaled_X, y)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X_selected,y,test_size=0.2,shuffle=True,random_state=123,stratify=y)
```

```python
X_train.shape , y_train.shape # Training sample
```

```python
X_test.shape, y_test.shape #Testing sample
```
